## Create Tables

```

CREATE TABLE Customers (
    CustomerID INT PRIMARY KEY,
    CustomerName VARCHAR(50),
    City VARCHAR(50)
);

CREATE TABLE Orders (
    OrderID INT PRIMARY KEY,
    CustomerID INT,
    OrderDate DATE,
    Amount INT
);

CREATE TABLE Payments (
    PaymentID VARCHAR(10) PRIMARY KEY,
    CustomerID INT,
    PaymentDate DATE,
    Amount INT
);

CREATE TABLE Employees (
    EmployeeID INT PRIMARY KEY,
    EmployeeName VARCHAR(50),
    ManagerID INT
);
```

## Q1

```
SELECT DISTINCT c.* FROM Customers c INNER JOIN Orders o ON c.CustomerID = o.CustomerID;
```

## Q2

```
SELECT c.*, o.* FROM Customers c LEFT JOIN Orders o ON c.CustomerID = o.CustomerID;
```

## Q3

```
SELECT o.*, c.* FROM Orders o LEFT JOIN Customers c ON o.CustomerID = c.CustomerID;
```

## Q4

```
SELECT c.*, o.* FROM Customers c LEFT JOIN Orders o ON c.CustomerID = o.CustomerID
UNION
SELECT c.*, o.* FROM Customers c RIGHT JOIN Orders o ON c.CustomerID = o.CustomerID;
```

## Q5

```
SELECT c.* FROM Customers c LEFT JOIN Orders o ON c.CustomerID = o.CustomerID WHERE o.CustomerID IS NULL;
```

## Q6

```
SELECT DISTINCT p.CustomerID FROM Payments p LEFT JOIN Orders o ON p.CustomerID = o.CustomerID WHERE o.CustomerID IS NULL;
```

## Q7

```
SELECT c.CustomerName, o.OrderID FROM Customers c CROSS JOIN Orders o;
```

## Q8

```
SELECT c.CustomerName, o.Amount AS OrderAmount, p.Amount AS PaymentAmount FROM Customers c LEFT JOIN Orders o ON c.CustomerID = o.CustomerID LEFT JOIN Payments p ON c.CustomerID = p.CustomerID;
```

## Q9

```
SELECT DISTINCT c.* FROM Customers c INNER JOIN Orders o ON c.CustomerID = o.CustomerID INNER JOIN Payments p ON c.CustomerID = p.CustomerID;
```
